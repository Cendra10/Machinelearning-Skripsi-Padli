```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity




# Path file CSV hasil TF-IDF
tfidf_file_path = '/content/gdrive/MyDrive/modelling/tfidf_matrix.csv'

# Memuat DataFrame hasil TF-IDF
tfidf_df = pd.read_csv(tfidf_file_path)

# Tampilkan beberapa baris pertama dari DataFrame
print(tfidf_df.head())


# Menentukan kategori yang ingin dielompokkan
kategori = ['atur', 'konten', 'kecewa']

# Membuat DataFrame untuk setiap kategori
kategori_dfs = [tfidf_df[tfidf_df['comment'].str.contains(kat)].reset_index(drop=True) for kat in kategori]

# Menghitung cosine similarity untuk setiap kategori
for i, kat_df in enumerate(kategori_dfs):
    cosine_sim_matrix = cosine_similarity(tfidf_df.loc[kat_df.index], tfidf_df.loc[kat_df.index])

    # Menentukan batas similarity untuk pengelompokan
    threshold = 0.6

    # Menambahkan kolom cluster ke DataFrame kategori
    kat_df['cluster'] = (cosine_sim_matrix > threshold).sum(axis=1)

    # Menampilkan hasil
    print(f"\nHasil pengelompokan untuk kategori '{kategori[i]}':")
    print(kat_df[['comment', 'cluster']])

    # Menampilkan jumlah klaster yang terbentuk
    print(f"Jumlah klaster yang terbentuk: {kat_df['cluster'].nunique()}")
```
