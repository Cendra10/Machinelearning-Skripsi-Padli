```python
  app_packages =[
      'com.google.android.apps.youtube.kids'
  ]
```

```python
app_infos = []

for application in tqdm(app_packages):
  info =app(application, lang='id', country='id')
  del info['comments']
  app_infos.append(info)
```

```python
app_reviews = []
for ap in tqdm(app_packages):
  for score in list(range(1,6)):
    for sort_order in [Sort.MOST_RELEVANT, Sort.NEWEST]:
      rvs, _= reviews(
          ap,
          lang='id',
          country = 'id',
          sort = sort_order,
          count=5000,
          filter_score_with = score
      )
      for r in rvs:
        r['sortOder'] = 'most_relevant' if sort_order == Sort.MOST_RELEVANT else 'newest'
        r['appId'] = ap
        app_reviews.extend(rvs)
```

```python
app_reviews_df = pd.DataFrame(app_reviews)

app_reviews_df = app_reviews_df.sort_values(["userName", "content", "score"])
```

```python
app_reviews_df.to_csv(path+'hasilreviews.csv', index=None, header=True)
```
